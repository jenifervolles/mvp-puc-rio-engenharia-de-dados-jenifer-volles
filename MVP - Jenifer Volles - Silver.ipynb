```sql
-- Retorna as linhas duplicadas: agrupa pelas 15 colunas que serão mantidas na Silver e mostra grupos com mais de 1 ocorrência

WITH duplicatas AS (
    SELECT
        EMPRESA_NOME, EMPRESA_NACIONALIDADE, ANO, MES,
        AEROPORTO_DE_ORIGEM_NOME, AEROPORTO_DE_ORIGEM_PAIS,
        AEROPORTO_DE_DESTINO_NOME, AEROPORTO_DE_DESTINO_PAIS,
        NATUREZA, GRUPO_DE_VOO,
        PASSAGEIROS_PAGOS, PASSAGEIROS_GRATIS,
        DISTANCIA_VOADA_KM, ASSENTOS, DECOLAGENS,
        ROW_NUMBER() OVER (
            PARTITION BY
                EMPRESA_NOME, EMPRESA_NACIONALIDADE, ANO, MES,
                AEROPORTO_DE_ORIGEM_NOME, AEROPORTO_DE_ORIGEM_PAIS,
                AEROPORTO_DE_DESTINO_NOME, AEROPORTO_DE_DESTINO_PAIS,
                NATUREZA, GRUPO_DE_VOO,
                PASSAGEIROS_PAGOS, PASSAGEIROS_GRATIS,
                DISTANCIA_VOADA_KM, ASSENTOS, DECOLAGENS
            ORDER BY ANO, MES, EMPRESA_NOME
        ) AS rn
    FROM workspace.default.dados_anac_bronze
)
SELECT
    EMPRESA_NOME, EMPRESA_NACIONALIDADE, ANO, MES,
    AEROPORTO_DE_ORIGEM_NOME, AEROPORTO_DE_ORIGEM_PAIS,
    AEROPORTO_DE_DESTINO_NOME, AEROPORTO_DE_DESTINO_PAIS,
    NATUREZA, GRUPO_DE_VOO,
    PASSAGEIROS_PAGOS, PASSAGEIROS_GRATIS,
    DISTANCIA_VOADA_KM, ASSENTOS, DECOLAGENS
FROM duplicatas
WHERE rn > 1
ORDER BY ANO, MES, EMPRESA_NOME;
```

```sql
-- Verifica quais colunas têm NULL e se os NULLs ocorrem apenas em voos INTERNACIONAIS

SELECT
    'AEROPORTO_DE_ORIGEM_NOME' AS coluna_null,
    COUNT(*) AS total_nulls,
    SUM(CASE WHEN NATUREZA = 'INTERNACIONAL' THEN 1 ELSE 0 END) AS internacionais,
    SUM(CASE WHEN NATUREZA = 'DOMÉSTICA' THEN 1 ELSE 0 END) AS domesticos
FROM workspace.default.dados_anac_silver WHERE AEROPORTO_DE_ORIGEM_NOME IS NULL
UNION ALL
SELECT
    'AEROPORTO_DE_ORIGEM_PAIS',
    COUNT(*),
    SUM(CASE WHEN NATUREZA = 'INTERNACIONAL' THEN 1 ELSE 0 END),
    SUM(CASE WHEN NATUREZA = 'DOMÉSTICA' THEN 1 ELSE 0 END)
FROM workspace.default.dados_anac_silver WHERE AEROPORTO_DE_ORIGEM_PAIS IS NULL
UNION ALL
SELECT
    'GRUPO_DE_VOO',
    COUNT(*),
    SUM(CASE WHEN NATUREZA = 'INTERNACIONAL' THEN 1 ELSE 0 END),
    SUM(CASE WHEN NATUREZA = 'DOMÉSTICA' THEN 1 ELSE 0 END)
FROM workspace.default.dados_anac_silver WHERE GRUPO_DE_VOO IS NULL
UNION ALL
SELECT
    'PASSAGEIROS_PAGOS',
    COUNT(*),
    SUM(CASE WHEN NATUREZA = 'INTERNACIONAL' THEN 1 ELSE 0 END),
    SUM(CASE WHEN NATUREZA = 'DOMÉSTICA' THEN 1 ELSE 0 END)
FROM workspace.default.dados_anac_silver WHERE PASSAGEIROS_PAGOS IS NULL
UNION ALL
SELECT
    'PASSAGEIROS_GRATIS',
    COUNT(*),
    SUM(CASE WHEN NATUREZA = 'INTERNACIONAL' THEN 1 ELSE 0 END),
    SUM(CASE WHEN NATUREZA = 'DOMÉSTICA' THEN 1 ELSE 0 END)
FROM workspace.default.dados_anac_silver WHERE PASSAGEIROS_GRATIS IS NULL
UNION ALL
SELECT
    'DISTANCIA_VOADA_KM',
    COUNT(*),
    SUM(CASE WHEN NATUREZA = 'INTERNACIONAL' THEN 1 ELSE 0 END),
    SUM(CASE WHEN NATUREZA = 'DOMÉSTICA' THEN 1 ELSE 0 END)
FROM workspace.default.dados_anac_silver WHERE DISTANCIA_VOADA_KM IS NULL
UNION ALL
SELECT
    'ASSENTOS',
    COUNT(*),
    SUM(CASE WHEN NATUREZA = 'INTERNACIONAL' THEN 1 ELSE 0 END),
    SUM(CASE WHEN NATUREZA = 'DOMÉSTICA' THEN 1 ELSE 0 END)
FROM workspace.default.dados_anac_silver WHERE ASSENTOS IS NULL;
```

```sql
-- Verificação completa do perfil das 5.212 linhas com origem NULL

SELECT
    COUNT(*) AS total_linhas_origem_null,
    SUM(CASE WHEN NATUREZA = 'INTERNACIONAL' THEN 1 ELSE 0 END) AS internacionais,
    SUM(CASE WHEN NATUREZA = 'DOMÉSTICA' THEN 1 ELSE 0 END) AS domesticos,
    SUM(CASE WHEN EMPRESA_NACIONALIDADE = 'ESTRANGEIRA' THEN 1 ELSE 0 END) AS empresas_estrangeiras,
    SUM(CASE WHEN EMPRESA_NACIONALIDADE = 'BRASILEIRA' THEN 1 ELSE 0 END) AS empresas_brasileiras,
    ROUND(SUM(CASE WHEN EMPRESA_NACIONALIDADE = 'ESTRANGEIRA' THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS pct_estrangeiras,
    SUM(CASE WHEN AEROPORTO_DE_DESTINO_NOME IS NOT NULL THEN 1 ELSE 0 END) AS destino_preenchido,
    SUM(CASE WHEN AEROPORTO_DE_DESTINO_NOME IS NULL THEN 1 ELSE 0 END) AS destino_null,
    SUM(CASE WHEN PASSAGEIROS_PAGOS IS NOT NULL THEN 1 ELSE 0 END) AS passageiros_pagos_preenchido,
    SUM(CASE WHEN PASSAGEIROS_PAGOS IS NULL THEN 1 ELSE 0 END) AS passageiros_pagos_null,
    SUM(CASE WHEN DISTANCIA_VOADA_KM IS NULL THEN 1 ELSE 0 END) AS distancia_null,
    SUM(CASE WHEN ASSENTOS IS NULL THEN 1 ELSE 0 END) AS assentos_null
FROM workspace.default.dados_anac_silver
WHERE AEROPORTO_DE_ORIGEM_NOME IS NULL;
```

```sql
-- Verificação de espaços em branco (leading/trailing) nas colunas de texto
-- Retorna contagem de linhas com espaços extras em cada coluna de texto

SELECT
    SUM(CASE WHEN EMPRESA_NOME != TRIM(EMPRESA_NOME) THEN 1 ELSE 0 END) AS espacos_EMPRESA_NOME,
    SUM(CASE WHEN EMPRESA_NACIONALIDADE != TRIM(EMPRESA_NACIONALIDADE) THEN 1 ELSE 0 END) AS espacos_EMPRESA_NACIONALIDADE,
    SUM(CASE WHEN AEROPORTO_DE_ORIGEM_NOME != TRIM(AEROPORTO_DE_ORIGEM_NOME) THEN 1 ELSE 0 END) AS espacos_ORIGEM_NOME,
    SUM(CASE WHEN AEROPORTO_DE_ORIGEM_PAIS != TRIM(AEROPORTO_DE_ORIGEM_PAIS) THEN 1 ELSE 0 END) AS espacos_ORIGEM_PAIS,
    SUM(CASE WHEN AEROPORTO_DE_DESTINO_NOME != TRIM(AEROPORTO_DE_DESTINO_NOME) THEN 1 ELSE 0 END) AS espacos_DESTINO_NOME,
    SUM(CASE WHEN AEROPORTO_DE_DESTINO_PAIS != TRIM(AEROPORTO_DE_DESTINO_PAIS) THEN 1 ELSE 0 END) AS espacos_DESTINO_PAIS,
    SUM(CASE WHEN NATUREZA != TRIM(NATUREZA) THEN 1 ELSE 0 END) AS espacos_NATUREZA,
    SUM(CASE WHEN GRUPO_DE_VOO != TRIM(GRUPO_DE_VOO) THEN 1 ELSE 0 END) AS espacos_GRUPO_DE_VOO
FROM workspace.default.dados_anac_silver
WHERE EMPRESA_NOME IS NOT NULL
   OR EMPRESA_NACIONALIDADE IS NOT NULL
   OR AEROPORTO_DE_ORIGEM_NOME IS NOT NULL
   OR AEROPORTO_DE_ORIGEM_PAIS IS NOT NULL
   OR AEROPORTO_DE_DESTINO_NOME IS NOT NULL
   OR AEROPORTO_DE_DESTINO_PAIS IS NOT NULL
   OR NATUREZA IS NOT NULL
   OR GRUPO_DE_VOO IS NOT NULL;
```

```sql
-- Verificação de strings vazias ('') nas colunas de texto
-- Retorna contagem de linhas com string vazia em cada coluna de texto

SELECT
    SUM(CASE WHEN EMPRESA_NOME = '' THEN 1 ELSE 0 END) AS vazia_EMPRESA_NOME,
    SUM(CASE WHEN EMPRESA_NACIONALIDADE = '' THEN 1 ELSE 0 END) AS vazia_EMPRESA_NACIONALIDADE,
    SUM(CASE WHEN AEROPORTO_DE_ORIGEM_NOME = '' THEN 1 ELSE 0 END) AS vazia_ORIGEM_NOME,
    SUM(CASE WHEN AEROPORTO_DE_ORIGEM_PAIS = '' THEN 1 ELSE 0 END) AS vazia_ORIGEM_PAIS,
    SUM(CASE WHEN AEROPORTO_DE_DESTINO_NOME = '' THEN 1 ELSE 0 END) AS vazia_DESTINO_NOME,
    SUM(CASE WHEN AEROPORTO_DE_DESTINO_PAIS = '' THEN 1 ELSE 0 END) AS vazia_DESTINO_PAIS,
    SUM(CASE WHEN NATUREZA = '' THEN 1 ELSE 0 END) AS vazia_NATUREZA,
    SUM(CASE WHEN GRUPO_DE_VOO = '' THEN 1 ELSE 0 END) AS vazia_GRUPO_DE_VOO
FROM workspace.default.dados_anac_silver;
```
