Abaixo segue query para criação da tabela bronze, com os dados fiéis ao arquivo .csv

```sql
%sql
create or replace table workspace.default.dados_anac_bronze
using delta
TBLPROPERTIES ('delta.columnMapping.mode' = 'name', 'delta.minReaderVersion' = '2', 'delta.minWriterVersion' = '5')
as select * from read_files('/Volumes/workspace/default/dados_anac', format => 'csv', sep => ';', header => true, quote => '"', skipRows => 1, inferSchema => true, schemaEvolutionMode => 'none')
```
